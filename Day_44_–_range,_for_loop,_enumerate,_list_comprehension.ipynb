```python
# Project Title: Employee Hours Report
# Scenario: You are analyzing student exam scores to generate performance reports.
# Tasks:
# 1. Create a list of 10 random exam scores ranging from 0 to 100 using list comprehension.
# 2. Display each score with its index using the enumerate function.
# 3. Find all scores greater than 75 and display them.
# 4. Calculate and display the average score.
```

```python
import random                                                                    # The underscore _ is a placeholder variable used when the loop variable is not needed.
                                                                                 # It indicates that the code should repeat 10 times without using the iteration value.
# 1. Create a list of 10 random exam scores (0–100) using list comprehension

scores = [random.randint(0, 100) for _ in range(10)]                             # random.randint(0, 100) generates random integers between 0 and 100.
                                                                                 # Generates numbers from 0 to 9.
# 2. Display each score with its index using enumerate

print("Exam Scores with Index:")
for index, score in enumerate(scores):                                           # enumerate(scores) pairs each score with its index.
    print(f"Index {index}: Score {score}")

# 3. Find all scores greater than 75

high_scores = [score for score in scores if score > 75]                          # List comprehension filters scores greater than 75.
print("\nScores greater than 75:", high_scores)

# 4. Calculate and display the average score

average_score = sum(scores) / len(scores)                                        # sum(scores) / len(scores) calculates the average.
print("\nAverage Score:", average_score)
```
