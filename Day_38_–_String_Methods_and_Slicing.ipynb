```python
# Project Title: Feedback Analyzer

# 1. Ask the user to input a sentence (feedback)

feedback = input("Please enter your feedback: ")

# 2. Count how many words are in the sentence

word_count = len(feedback.split())                                         # .split() - Splits the string into a list of words, using spaces as the default separator.
print(f"\nTotal number of words: {word_count}")                            # len(...) - Counts how many items are in that list.
                                                                           # fstring - embed variables or expressions directly inside a string.
# 3. Check how many times the word “data” appears (case-insensitive)

data_count = feedback.lower().count("data")                                # .lower() - Converts the entire string to lowercase.
print(f"Occurrences of the word 'data': {data_count}")                     # .count("data") - Counts how many times the substring appears in the lowercase string.

# 4. Extract and print the first 5 and last 5 characters

print(f"First 5 characters: {feedback[0:5]}")        # : is the slice operator. [:5] means: start at index 0 (the beginning) and go up to, but not including, index 5.
print(f"Last 5 characters: {feedback[-5:]}")         # -1 → last character, -2 → second last character, feedback[-5:] → last 5 characters (indexes -5 to -1).

# 5. Reverse the sentence and print it

print(f"Reversed feedback: {feedback[::-1]}")        # start → not given, end → not given, step = -1 → means move backwards one character at a time.
```
