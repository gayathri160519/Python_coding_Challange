```python
# Global variable                                                                # A global variable is declared outside the class and functions.

transaction_count = 0                                                            # It can be accessed from anywhere in the program.


# Class definition

class BankAccount:                                                               # A class acts as a blueprint for creating bank account objects.

    def __init__(self, account_holder, balance=0):                               # The constructor runs automatically when an object is created.
        self.account_holder = account_holder                                     # stores the values inside the object.
        self.balance = balance

    def deposit(self, amount):                                                   # Adds money to the account.
        global transaction_count                                                 # This tells Python to use the global variable instead of creating a local one.

        # Local variable

        new_balance = self.balance + amount                                      # new_balance is a local variable because it exists only inside the method.

        self.balance = new_balance                                               # means the balance of the current account object.
        transaction_count += 1

        print(f"₹{amount} deposited successfully.")

    def withdraw(self, amount):
        global transaction_count

        if amount > self.balance:                                                # Checks whether sufficient balance exists.
            print("Insufficient balance! Withdrawal denied.")                    # This prevents negative balances.
        else:
            self.balance -= amount
            transaction_count += 1
            print(f"₹{amount} withdrawn successfully.")

    def display_balance(self):                                                   # Displays account details.
        print(f"Account Holder: {self.account_holder}")
        print(f"Current Balance: ₹{self.balance}")


# Create account object                                                          # An object is an instance of a class.

account1 = BankAccount("Gayathri", 5000)

# Positional argument

account1.deposit(1500)                                                           # The value is passed based on its position.

# Keyword argument

account1.withdraw(amount=2000)                                                   # The parameter name is explicitly specified.

# Attempt to withdraw more than available balance

account1.withdraw(amount=10000)

# Display final balance

account1.display_balance()

# Display total transactions

print("Total Transactions:", transaction_count)
```
