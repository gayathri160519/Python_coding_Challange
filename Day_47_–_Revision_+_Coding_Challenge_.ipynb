```python
import json                                                                      # JSON is used to store and exchange data in a structured format.

# Constant data using tuple

AVAILABLE_CLASSES = ("Python", "SQL", "Power BI", "Data Analytics")              # Tuples are immutable.

# Student data storage                                                           # Each student is stored as a dictionary.

students = []                                                                    # List -> Stores multiple students.


# Function to categorize performance

def get_performance(avg):
    if avg >= 90:                                                                # Used to categorize performance.
        return "Excellent"
    elif avg >= 75:
        return "Good"
    elif avg >= 50:
        return "Average"
    else:
        return "Needs Improvement"


# Add student

def add_student():                                                               # Functions help reuse code.
    try:                                                                         # Exception Handling -> Prevents program crashes due to invalid user input.
        name = input("Enter student name: ")
        age = int(input("Enter age: "))
        grades = list(map(int, input("Enter grades separated by spaces: ").split()))  # split() -> It breaks a string into pieces using spaces by default.
                                                                                      # map() applies a function to each element of an iterable.
        student = {
            "name": name,
            "age": age,
            "grades": grades
        }

        students.append(student)
        print("Student added successfully!")

    except ValueError:
        print("Invalid input! Please enter numeric values.")


# Display students

def display_students():
    if not students:
        print("No student records found.")
        return

    for student in students:
        avg = sum(student["grades"]) / len(student["grades"])
        performance = get_performance(avg)

        print("\nName:", student["name"])
        print("Age:", student["age"])
        print("Grades:", student["grades"])
        print("Average:", round(avg, 2))
        print("Performance:", performance)


# Search student

def search_student():
    name = input("Enter student name to search: ")

    for student in students:
        if student["name"].lower() == name.lower():                              # Sorts students by highest average marks first.
            print(student)
            return

    print("Student not found.")


# Update student

def update_student():
    name = input("Enter student name to update: ")

    for student in students:
        if student["name"].lower() == name.lower():

            student["age"] = int(input("Enter new age: "))
            student["grades"] = list(
                map(int, input("Enter new grades: ").split())
            )

            print("Student record updated!")
            return

    print("Student not found.")


# Delete student

def delete_student():
    name = input("Enter student name to delete: ")

    for student in students:
        if student["name"].lower() == name.lower():
            students.remove(student)
            print("Student deleted successfully!")
            return

    print("Student not found.")


# Sort students by average grade

def sort_students():
    sorted_students = sorted(
        students,
        key=lambda s: sum(s["grades"]) / len(s["grades"]),
        reverse=True
    )

    print("\nStudents Sorted by Average Marks")

    for student in sorted_students:
        avg = sum(student["grades"]) / len(student["grades"])
        print(student["name"], "-", round(avg, 2))


# Filter top performers

def filter_top_students():

    print("\nTop Performers (Average > 75)")

    for student in students:
        avg = sum(student["grades"]) / len(student["grades"])

        if avg > 75:                                                             # Shows only high-performing students.
            print(student["name"], "-", round(avg, 2))


# Save data to file

def save_data():
    with open("students.json", "w") as file:                                     # File Handling : To Save
        json.dump(students, file)

    print("Data saved successfully.")


# Load data from file

def load_data():
    global students

    try:                                                                         # To Load.
        with open("students.json", "r") as file:                                 # Data remains available even after closing the program.
            students = json.load(file)

        print("Data loaded successfully.")

    except FileNotFoundError:
        print("No saved file found.")


# Main Menu

while True:

    print("\n===== STUDENT MANAGEMENT SYSTEM =====")
    print("1. Add Student")
    print("2. Display Students")
    print("3. Search Student")
    print("4. Update Student")
    print("5. Delete Student")
    print("6. Sort Students")
    print("7. Filter Top Students")
    print("8. Save Data")
    print("9. Load Data")
    print("10. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        display_students()

    elif choice == "3":
        search_student()

    elif choice == "4":
        update_student()

    elif choice == "5":
        delete_student()

    elif choice == "6":
        sort_students()

    elif choice == "7":
        filter_top_students()

    elif choice == "8":
        save_data()

    elif choice == "9":
        load_data()

    elif choice == "10":
        print("Exiting program...")
        break

    else:
        print("Invalid choice!")
```
