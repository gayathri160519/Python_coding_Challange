```python
# Project Title: Retail Inventory Manager
# Scenario: You are managing an inventory system for a retail shop. Certain product categories are fixed, but the actual products can be added or removed by the staff.


# 1. Create a tuple with fixed product categories

categories = ("Electronics", "Stationery", "Clothing")                           # Tuple (categories) → Used for fixed product categories since tuples are immutable.

# 2. Create a list of products that can be updated dynamically

products = ["Laptop", "Notebook", "T-Shirt"]                                     # List (products) → Used for products because lists are dynamic and allow adding/removing items.

# 3. Ask the user to enter a new product and add it to the list

new_product = input("Enter a new product to add: ")
products.append(new_product)                                                     # append() → Adds a new product.

# 4. Ask the user to input the index of a product to remove it from the list

print("Current products:", products)
remove_index = int(input("Enter the index of the product to remove: "))

# Check if index is valid

if 0 <= remove_index < len(products):                                            # 0 <= remove_index → ensures the index is not negative.
  removed_product = products.pop(remove_index)                                   # remove_index < len(products) → ensures the index is less than the length of the list (so it exists).
  print(f"Removed product: {removed_product}")                                   # pop(index) → Removes a product at the given index.
else:
    print("Invalid index! No product removed.")

# 5. Display the final list of products and the product categories

print("\nFinal Products List:", products)
print("Fixed Product Categories:", categories)
```
