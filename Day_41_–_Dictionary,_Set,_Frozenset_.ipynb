```python
# Project Title: Student Grades & Hobbies Tracker
# Scenario: A student activity coordinator is collecting information about student grades and hobbies. They want to ensure grades are updated properly and hobby lists are stored efficiently.

# 1. Create a dictionary with student names as keys and grades as values

student_grades = {                                                               # student_grades dictionary stores names as keys and grades as values.
    "Aravind": 85,
    "Banupriya": 72,
    "Chandru": 90
}

# Function to display grades

def display_grades():                                                            # To Avoid repetition. With def, you write it once and reuse it.
    print("\n Student Grades")                                                   # It is modular, clean, and efficient.
    for name, grade in student_grades.items():
        print(f"{name}: {grade}")


# 2. Allow user to add or update grades for at least 3 students

for i in range(3):                                                               # Loop runs 3 times, letting the user add/update grades. If the name exists, the grade is updated; if not, a new entry is added.
    name = input("Enter student name: ")
    grade = int(input(f"Enter grade for {name}: "))
    student_grades[name] = grade

display_grades()

# 3. Create a set of hobbies and allow user to input at least 3 hobbies

hobbies = set()                                                                  #  creates an empty set.
for i in range(3):                                                               # A set is used for hobbies (no duplicates allowed). User inputs at least 3 hobbies.
    hobby = input("Enter a hobby: ")
    hobbies.add(hobby)                                                           # inserts a new item into the set. because they only store unique values and don’t care about order.

print("\nHobbies Set:", hobbies)

# 4. Convert the set of hobbies into a frozenset and display it
                                                                                 # Frozenset → Immutable (fixed, cannot be changed). unique, unordered values.
frozen_hobbies = frozenset(hobbies)                                              # frozenset makes the hobbies immutable (cannot be changed later).
print("Frozen Hobbies:", frozen_hobbies)
```
