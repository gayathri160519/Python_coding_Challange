```python
# Functions for basic operations

def add(a, b):                                                                   # Creates a function named add() that returns the sum of two numbers.
    return a + b                                                                 # subtract() → subtraction, multiply() → multiplication, divide() → division.

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error! Division by zero is not allowed."
    return a / b

# User input

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Enter operation (+, -, *, /): ")

# Perform operation

if operation == "+":
    result = add(num1, num2)

elif operation == "-":
    result = subtract(num1, num2)

elif operation == "*":
    result = multiply(num1, num2)

elif operation == "/":
    result = divide(num1, num2)

else:
    result = "Invalid operation!"

# Display result

print("\nResult:", result)

# Built-in functions demonstration

if isinstance(result, (int, float)):                                             # It checks whether result is a numeric value before applying abs(), round(), and pow().
    print("Absolute Value:", abs(result))                                        # Returns the absolute (positive) value.
    print("Rounded Value:", round(result, 2))                                    # Rounds the number to 2 decimal places.
    print("Power (result²):", pow(result, 2))                                    # Calculates the power.8
```
