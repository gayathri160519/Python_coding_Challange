```python
# Project Title: Contact Directory
# Scenario: You have been assigned to build a contact management system where employees can store and retrieve contact details efficiently.

# 1. Create a 2D list with sample contacts

contacts = [                                                                     # A 2D list (two-dimensional list) is simply a list inside another list.
    ["Alice", "9876543210", "alice@example.com"],
    ["Bob", "9123456780", "bob@example.com"],                                    # contacts[0] → ["Alice", "9876543210", "alice@example.com"]
    ["Charlie", "9988776655", "charlie@example.com"]                             # contacts[1] → ["Bob", "9123456780", "bob@example.com"]
]                                                                                # contacts[2] → ["Charlie", "9988776655", "charlie@example.com"]


def display_contacts():                                                          # This defines a function named display_contacts. Without def, we have to copy-paste the printing loop every time you want to show contacts.
                                                                                 # With def, we just call.

    print("\n Contact List")
    for contact in contacts:                                                     # Using for loop to print each contact.
        print(f"Name: {contact[0]}, Phone: {contact[1]}, Email: {contact[2]}")


# 2. Add a new contact

new_contact = ["David", "9001122334", "david@example.com"]
contacts.append(new_contact)                                                     # append() adds a new contact.

# 3. Remove the last item

contacts.pop()                                                                   # pop() removes the last contact.

# 4. Get new items from the user

name = input("Enter contact name: ")                                             # input() lets the user add new details dynamically.
phone = input("Enter phone number: ")
email = input("Enter email address: ")
contacts.append([name, phone, email])

# 5. Sort the list in ascending order based on names

contacts.sort(key=lambda x: x[0])                                                # sort() arranges contacts alphabetically by name.
                                                                                 # key= → tells Python how to compare each row in the 2D list.
                                                                                 # lambda x: x[0] → is a small anonymous function that says: Take each row (x).
                                                                                 # Use the first element (x[0], which is the Name) as the sorting key.
                                                                                 # # So the list gets sorted alphabetically by Name.
# Display final contacts

display_contacts()                                                               # whole block runs automatically.
                                                                                 # Used AI Support.
```
