```python
# Day 42 – Operators, Conditional Statements
# Task: Given an integer, n, perform the following conditional actions:
# ● If n is odd, print Weird
# ● If  n is even and in the inclusive range of  2 to 5, print Not Weird
# ● If n  is even and in the inclusive range of  6 to 20, print Weird
# ● If n  is even and greater than 20, print Not Weird
# Input Format :A single line containing a positive integer,n.
# Constraints ● 1<=n <=100
# Output Format: Print Weird if the number is weird. Otherwise, print Not Weird.

n = int(input("Enter a positive integer: "))

if n % 2 != 0:                                                                   # If this is True, Python runs this block and skips everything else.
    print("Weird")                                                               # if is False, Python moves on to check the next condition using elif:
elif n % 2 == 0 and 2 <= n <= 5:                                                 # If the number is even and between 2 and 5, then print Not Weird.
    print("Not Weird")
elif n % 2 == 0 and 6 <= n <= 20:                                                # If that’s not true, Python checks the next elif:
    print("Weird")                                                               # If the number is even and between 6 and 20, then print Weird.
elif n % 2 == 0 and n > 20:
    print("Not Weird")
```
