```python
# Project Title: Shopping Cart System
# Scenario: You are creating a simple billing system where customers can enter the prices of items they want to purchase. The program should keep accepting prices until the user decides to stop, then apply appropriate discounts before generating the final bill.
# Tasks:
# ● Continuously accept item prices using a while loop.
# ● Stop taking input when the user enters 0 (sentinel value).
# ● Keep track of the total number of items and the subtotal amount.
# ● Ensure no negative prices are allowed (ask again without counting as an item).
# ● Apply discounts based on the subtotal:
# ○ 20% for totals ≥ 1000
# ○ 10% for totals ≥ 500
# ○ 5% for totals ≥ 200
# ○ 0% for totals < 200
# ● Display a clear bill summary including subtotal, discount amount, and final total.
```

```python
# Shopping Cart Billing System

subtotal = 0
item_count = 0

while True:                                                                        # Keeps asking for item prices until you enter 0.
    try:
        price = float(input("Enter item price (0 to finish): "))

        if price == 0:
            break
        elif price < 0:
            print("Invalid price! Please enter a positive value.")
            continue
        else:
            subtotal += price                                                    #  running total of all item prices entered
            item_count += 1                                                      # It increases the count of items by one each time a valid price is entered.
    except ValueError:
        print(" Invalid input! Please enter a number.")

                                                                                 # Apply discounts
if subtotal >= 1000:
    discount_rate = 0.20
elif subtotal >= 500:
    discount_rate = 0.10
elif subtotal >= 200:
    discount_rate = 0.05
else:
    discount_rate = 0.0

discount_amount = subtotal * discount_rate                                       # This calculates how much money the customer saves due to the discount.
final_total = subtotal - discount_amount                                         # This calculates the actual amount the customer has to pay after subtracting the discount.

                                                                                 # Bill Summary
print("\n Bill Summary")
print("Total items:", item_count)
print("Subtotal: ₹", subtotal)
print("Discount applied:", int(discount_rate * 100), "%")                        # converting the decimal discount rate into a percentage value for display.
print("Discount amount: ₹", discount_amount)
print("Final total: ₹", final_total)
print(" Thank you for shopping with us!")
```
